```python
import sys
from pathlib import Path

PROJECT_ROOT = Path(r"D:\plant_fl_project")
sys.path.append(str(PROJECT_ROOT))
```

```python
from shared.model import build_model, compile_model
from shared.preprocessing import (
    load_labels,
    make_train_val_datasets,
    compute_class_weights,
)
from shared.config import RAW_DATA_DIR, LABELS_FILE, NUM_CLASSES, EPOCHS
```

```python
labels = load_labels(LABELS_FILE)

train_ds, val_ds, class_names = make_train_val_datasets(RAW_DATA_DIR)
```

```python
class_weights, counts = compute_class_weights(RAW_DATA_DIR, class_names=class_names)
```

```python
model = build_model(num_classes=NUM_CLASSES)
model = compile_model(model)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    class_weight=class_weights
)
```
