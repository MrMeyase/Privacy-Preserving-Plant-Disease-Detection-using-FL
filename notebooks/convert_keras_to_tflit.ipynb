```python
import tensorflow as tf

# Load trained model
model = tf.keras.models.load_model(r"D:\plant_fl_project\server\checkpoints\global_model.keras")

# Convert to TFLite
converter = tf.lite.TFLiteConverter.from_keras_model(model)

# Optional optimization (VERY IMPORTANT for Pi)
converter.optimizations = [tf.lite.Optimize.DEFAULT]

tflite_model = converter.convert()

# Save model
with open(r"D:\plant_fl_project\android\AndroidStudioProject\app\src\main\assets\model.tflite", "wb") as f:
    f.write(tflite_model)

print("✅ TFLite model saved!")
```
